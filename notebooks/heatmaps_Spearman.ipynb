# Тепловая карта: коэффициент Спирмена (регрессия)

Тот же формат, что у `heatmap_regression_MAE` из `heatmaps_MAE_AUC90.ipynb`: панели горизонтально в порядке
**Random → Catalyst (GroupKFold-10) → LORGO**, в ячейке — точечная оценка и под ней ± половина ширины 95%
доверительного интервала (кластерный бутстреп). Dummy не показан.

Данные — `results/metrics_summary.csv` из регрессионного ноутбука (MAE, RMSE, R², ρ Спирмена и
τ Кендалла, рассчитанные по тем же out-of-fold предсказаниям и тем же бутстреп-повторам). Переключатель `METRIC`
строит ту же карту для τ Кендалла.

Цвет: `RdBu` с центром 0 и общей симметричной шкалой для трёх панелей (синий — положительная корреляция).

**Оговорка для LORGO.** Метрики считаются по предсказаниям, объединённым по фолдам. При исключении целой группы
среднее обучающей выборки сдвигается против уровня исключённой группы, поэтому даже Dummy получает ρ = −0,54.
Объединённый ρ на панели LORGO включает этот сдвиг и не равен качеству ранжирования внутри группы.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm

# ----------------------------- CONFIGURATION -----------------------------
REG_CSV = Path("../results/regression/metrics_summary.csv")   # relative to notebooks/
OUT_DIR = Path("../results/figures/figures_heatmaps_spearman")

METRIC = "Spearman_rho"       # "Spearman_rho" или "Kendall_tau"
METRIC_LABELS = {"Spearman_rho": "Spearman ρ", "Kendall_tau": "Kendall τ"}
DPI_PNG = 600

FIG_WIDTH_MM  = 180
FIG_HEIGHT_MM = 125
FS_VALUE = 7
FS_CI    = 5.5
FS_TICK  = 7
FS_TITLE = 7.5
CMAP = "RdBu"
# -------------------------------------------------------------------------

SPLITS = ["random", "gkf10_catalyst", "lorgo_group"]
SPLIT_LABELS = {"random": "Random (5-fold × 5)",
                "gkf10_catalyst": "GroupKFold-10 (catalyst)",
                "lorgo_group": "LORGO (research group)"}
REP_ORDER = ["physchem", "maccs", "ecfp4", "ecfp4_count", "drfp", "onehot", "dft", "cgr"]
REP_LABELS = {"physchem": "Physchem", "maccs": "MACCS", "ecfp4": "ECFP4", "ecfp4_count": "ECFP4-count",
              "drfp": "DRFP", "onehot": "OHE", "dft": "DFT", "cgr": "CGR + catalyst"}
REG_MODELS = ["ridge", "svr", "gp", "rf", "xgb", "chemprop"]
MODEL_LABELS = {"ridge": "Ridge", "svr": "SVR", "gp": "GP", "rf": "RF", "xgb": "XGBoost", "chemprop": "D-MPNN"}

MM = 1 / 25.4
MISSING = "#f0f0f0"
TEXT_DARK = "#222222"
OUT_DIR.mkdir(parents=True, exist_ok=True)

mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Inter", "Arial", "Helvetica", "DejaVu Sans"],
    "pdf.fonttype": 42, "svg.fonttype": "none",
})

reg = pd.read_csv(REG_CSV)
reg = reg[reg.model != "dummy"]
assert METRIC in reg.columns, f"{METRIC} not in {REG_CSV}"
print("регрессия:", reg.shape, "| метрика:", METRIC)
```

## Вспомогательные функции

```python
def grid(df, col, models):
    """Матрица «представление × модель» для одного столбца."""
    return (df.pivot(index="representation", columns="model", values=col)
              .reindex(index=REP_ORDER, columns=models).to_numpy(float))


def heat(ax, df, metric, models, cmap, norm, title, ylabels=True):
    M = grid(df, metric, models)
    half = (grid(df, f"{metric}_ci_high", models) - grid(df, f"{metric}_ci_low", models)) / 2
    im = ax.imshow(np.ma.masked_invalid(M), cmap=cmap, norm=norm, aspect="auto")
    ax.set_facecolor(MISSING)
    for (i, j), v in np.ndenumerate(M):
        if np.isnan(v):
            continue
        r, g, b, _ = im.cmap(im.norm(v))
        col = "white" if 0.2126 * r + 0.7152 * g + 0.0722 * b < 0.5 else TEXT_DARK
        txt = f"{v:.2f}".replace("-0.00", "0.00")       # no negative zero
        ax.text(j, i - 0.14, txt, ha="center", va="center", fontsize=FS_VALUE, color=col)
        ax.text(j, i + 0.22, f"± {half[i, j]:.2f}", ha="center", va="center", fontsize=FS_CI, color=col)
    ax.set_xticks(range(len(models)), [MODEL_LABELS[m] for m in models], rotation=40, ha="right",
                  rotation_mode="anchor")
    ax.set_yticks(range(len(REP_ORDER)), [REP_LABELS[r] for r in REP_ORDER] if ylabels else [])
    ax.tick_params(length=0, labelsize=FS_TICK, pad=2)
    for s in ax.spines.values():
        s.set_visible(False)
    ax.set_title(title, fontsize=FS_TITLE, pad=4)
    return im


def colorbar(fig, im, ax, label=None, shrink=1.0, aspect=30):
    cb = fig.colorbar(im, ax=ax, orientation="horizontal", fraction=0.045, pad=0.02, aspect=aspect, shrink=shrink)
    cb.ax.xaxis.set_major_locator(mpl.ticker.MaxNLocator(5))
    cb.ax.tick_params(labelsize=FS_TICK - 1, length=2, pad=1)
    cb.outline.set_visible(False)
    if label:
        cb.set_label(label, fontsize=FS_TICK)
    return cb


def save(fig, stem):
    for ext in ("pdf", "svg"):
        fig.savefig(OUT_DIR / f"{stem}.{ext}", bbox_inches="tight")
    fig.savefig(OUT_DIR / f"{stem}.png", dpi=DPI_PNG, bbox_inches="tight")
```

## Регрессия: ранговая корреляция

```python
lim = float(np.nanmax(np.abs(reg[METRIC])))              # общая симметричная шкала для трёх схем
norm = TwoSlopeNorm(vcenter=0, vmin=-lim, vmax=lim)

fig, axes = plt.subplots(1, 3, figsize=(FIG_WIDTH_MM * MM, FIG_HEIGHT_MM * MM), constrained_layout=True)
for k, (ax, s) in enumerate(zip(axes, SPLITS)):
    im = heat(ax, reg[reg.split == s], METRIC, REG_MODELS, CMAP, norm, SPLIT_LABELS[s], ylabels=(k == 0))
colorbar(fig, im, list(axes), METRIC_LABELS[METRIC], shrink=0.5, aspect=40)
save(fig, f"heatmap_regression_{METRIC}")
plt.show()
```
